```python
#Create a clean copy
df_clean = df.copy()

df_clean.columns = [
    "State",
    "District",
    "Year",
    "Area_Hectare",
    "Production_Tonnes",
    "Yield_Tonne_Hectare"
]
```

```python
#Check Data Types
df_clean.dtypes
```

```python
#Check Missing Values
df_clean.isnull().sum()
```

```python
#Locate the missing records
df_clean[df_clean["Production_Tonnes"].isnull()]
```

```python
#Confirm the number of affected records
df_clean[df_clean["Production_Tonnes"].isnull()].shape
```

```python
#Check for duplicate records
df_clean.duplicated(
    subset=["State", "District", "Year"]
).sum()
```

```python
#Checking for negative numerical values
(df_clean[[
    "Area_Hectare",
    "Production_Tonnes",
    "Yield_Tonne_Hectare"
]] < 0).sum()
```

```python
#Count zero values
(df_clean[[
    "Area_Hectare",
    "Production_Tonnes",
    "Yield_Tonne_Hectare"
]] == 0).sum()
```

```python
#Locate all zero-production records
df_clean[df_clean["Production_Tonnes"] == 0]
```

```python
#Locate all zero-yield records
df_clean[df_clean["Yield_Tonne_Hectare"] == 0]
```

```python
#Find zero yield with reported production
df_clean[
    (df_clean["Yield_Tonne_Hectare"] == 0) &
    (df_clean["Production_Tonnes"] > 0)
]
```

```python
#Check the numerical distributions
df_clean[[
    "Area_Hectare",
    "Production_Tonnes",
    "Yield_Tonne_Hectare"
]].describe()
```

```python
#Find unusually high yields
df_clean.nlargest(
    10,
    "Yield_Tonne_Hectare"
)
```

```python
#Verify reported yield calculations
import numpy as np

df_clean["Calculated_Yield"] = (
    df_clean["Production_Tonnes"] / df_clean["Area_Hectare"]
)

df_clean["Yield_Difference"] = (
    df_clean["Yield_Tonne_Hectare"] - df_clean["Calculated_Yield"]
)

inconsistent = df_clean[
    df_clean["Production_Tonnes"].notna() &
    (df_clean["Yield_Difference"].abs() > 0.02)
]

print("Records with inconsistent yield:", len(inconsistent))
print(inconsistent[
    ["State", "District", "Year", "Production_Tonnes",
     "Area_Hectare", "Yield_Tonne_Hectare", "Calculated_Yield"]
].head(20))
```

```python
#Final Audit Summary
print("AGRICULTURE DATASET AUDIT")
print("-------------------------")
print("Total records:", len(df_clean))
print("Total columns:", len(df_clean.columns))
print("Unique states:", df_clean["State"].nunique())
print("Unique districts:", df_clean["District"].nunique())
print("Unique agricultural years:", df_clean["Year"].nunique())
print("Missing production:", df_clean["Production_Tonnes"].isna().sum())
print("Duplicate district-year records:",
      df_clean.duplicated(
          subset=["State", "District", "Year"]
      ).sum())
print("Negative numerical values:",
      (df_clean.select_dtypes(include="number") < 0).sum().sum())
print("Zero production:", (df_clean["Production_Tonnes"] == 0).sum())
print("Zero yield:", (df_clean["Yield_Tonne_Hectare"] == 0).sum())
print("Inconsistent yield records:", len(inconsistent))
```

```python
#Save extracted Dataset
from pathlib import Path

output_path = Path("../data/processed/agriculture_clean.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)

df_clean.to_csv(output_path, index=False)

print("Saved to:", output_path.resolve())
```
